# 02 — Feature Engineering

Build and validate all derived features used by the training pipeline.

```python
import sys
sys.path.insert(0, '../..')

import pandas as pd
import numpy as np
from training.preprocessing import engineer_features, ALL_FEATURES

df_raw = pd.read_csv('../../data/hdb_resale_before_2025.csv')
df = engineer_features(df_raw)
print('Features after engineering:')
df[ALL_FEATURES + ['log_resale_price']].describe()
```

```python
# Validate storey_midpoint parsing
print('storey_range samples:')
print(df_raw['storey_range'].value_counts().head(10))
print('\nstorey_midpoint range:', df['storey_midpoint'].min(), '–', df['storey_midpoint'].max())
```

```python
# Validate remaining_lease_years parsing
print('remaining_lease samples:')
print(df_raw['remaining_lease'].value_counts().head(10))
print('\nremaining_lease_years range:', df['remaining_lease_years'].min(), '–', df['remaining_lease_years'].max())
```

```python
# Correlation heatmap
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = ['floor_area_sqm', 'storey_midpoint', 'remaining_lease_years',
                'lease_commence_date', 'transaction_year', 'transaction_month', 'log_resale_price']
corr = df[numeric_cols].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```
